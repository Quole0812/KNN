````python
from keras.datasets import mnist
df = mnist.load_data('minst.db')
train,test = df
X_train, y_train = train
X_test, y_test = test

import matplotlib.pyplot as plt
import numpy as np
np.set_printoptions(edgeitems=30, linewidth=100000)
for i in range(5):
    print(y_test[i],'\n')
    print(X_test[i],'\n')
    plt.contourf(np.rot90(X_test[i].transpose()))
    plt.show()
````

````python
#2 
print("X_train shape")
print(X_train.shape)

print("X_test shape")
print(X_test.shape)

print("X_train[0] shape")
print(X_train[0].shape)

print("X_test[0] shape")
print(X_test[0].shape)

print("y_train shape")
print(y_train.shape)

print("y_test shape")
print(y_test.shape)
````

````python
#3 
X_train_flat = X_train.reshape(X_train.shape[0], 784)
X_test_flat = X_test.reshape(X_test.shape[0], 784)
````

````python
from sklearn.neighbors import KNeighborsClassifier
k_values = range(1, 11)
accuracies = []

for k in k_values:
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train_flat, y_train)
    y_hat = model.predict(X_test_flat)
    acc = np.sum(y_hat == y_test) / len(y_test)
    accuracies.append(acc)
    print(f"k={k}: Accuracy = {acc}")

# Find optimal k
optimal_k = k_values[np.argmax(accuracies)]
print(f"\nOptimal k: {optimal_k}")

plt.plot(k_values, accuracies)
plt.xlabel("k")
plt.ylabel("Accuracy")
plt.title("Accuracy vs k for kNN Classifier")
plt.show()
````

````python
#5
import pandas as pd

model = KNeighborsClassifier(n_neighbors=optimal_k)
model.fit(X_train_flat, y_train)  
y_hat = model.predict(X_test_flat)

accuracy = np.sum(y_hat == y_test) / len(y_test)
print(f"Accuracy: {accuracy:.4f} ({accuracy*100:.2f}%)")

confusion = pd.crosstab(y_test, y_hat, rownames=['Actual'], colnames=['Predicted'])
print("\nConfusion Matrix:")
print(confusion)
````
